```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
import transformers
```

### Data Processing

Load data

```python
# df = pd.read_csv('labeledTrainData.tsv.zip', sep='\t')

df = pd.read_csv('/kaggle/input/datasets/mashwinsrivaths/testds/train_(1)_(2)_(1).csv')



print(df.shape)
```

```python
#Sentences and labels
sentences = df.Review.values
labels = df.Rating.values
```

```python
sentences = df.apply(lambda x: x.Review_Title + " => " + x.Review, axis=1).values
```

## Tokenize data using Bert Tokenizer

```python
# from transformers import AutoTokenizer

# Load tokenizer (AutoTokenizer is the modern way)
tokenizer = transformers.AutoTokenizer.from_pretrained('bert-base-uncased')

```

```python
#tokenizer.vocab.items()
```

```python
tokenized_texts = [tokenizer.tokenize(sent) for sent in sentences]
```

```python
#We will use only first 200 tokens to do classification (this value can be changed)
max_length = 200
tokenized_texts = [sent[:max_length] for sent in tokenized_texts]
```

```python
for i in range(len(tokenized_texts)):
    sent = tokenized_texts[i]
    sent = ['[CLS]'] + sent + ['[SEP]']
    tokenized_texts[i] = sent
```

```python
print(tokenized_texts[0])
```

```python
#Convert tokens into IDs
input_ids = [tokenizer.convert_tokens_to_ids(sent) for sent in tokenized_texts]
```

```python
print(input_ids[0])
```

```python
#Pad our tokens which might be less than max_length size
input_ids = tf.keras.preprocessing.sequence.pad_sequences(input_ids,
                                                          maxlen=max_length+2,
                                                          truncating='post',
                                                          padding='post')
```

Split data between training and test

```python
#80% data will be used for training while 20% will be used for test
trainX, testX, trainY, testY = train_test_split(input_ids, labels,
                                                test_size=0.2, random_state=12345)
```

Create Attention masks : Attention masks are useful to ignore padding tokens. Mask value will be set to 0 for padding tokens and 1 for actual tokens. We will create mask both for training and test data

```python
# Create attention masks for training
train_attn_masks = []

# Create a mask of 1s for each token followed by 0s for padding
for seq in trainX:
  seq_mask = [float(i>0) for i in seq]
  train_attn_masks.append(seq_mask)
```

```python
# Create attention masks for Test
test_attn_masks = []

# Create a mask of 1s for each token followed by 0s for padding
for seq in testX:
  seq_mask = [float(i>0) for i in seq]
  test_attn_masks.append(seq_mask)
```

```python
print(train_attn_masks[100])
```

### Build Model

```python
# from transformers import TFBertForSequenceClassification

model = transformers.TFBertForSequenceClassification.from_pretrained(
    'bert-base-uncased',
    num_labels=2,
    from_pt=True   # explicitly convert from PyTorch
)
```

```python
model.summary()
```

```python
import tf_keras
# Use tf_keras optimizer to ensure compatibility with TFBert models in newer TF versions
optimizer = tf_keras.optimizers.Adam(learning_rate=3e-5, epsilon=1e-08, clipnorm=1.0)
```

```python
# Define loss and metrics using tf_keras
loss = tf_keras.losses.SparseCategoricalCrossentropy(from_logits=True)
metrics = [tf_keras.metrics.SparseCategoricalAccuracy('accuracy')]
```

```python
# !pip install "tensorflow<2.16"
```

```python
# 2. Compile the model using tf_keras compatible objects
# This avoids the '_distribute_strategy' error in TF 2.16+
model.compile(optimizer=optimizer, loss=loss, metrics=metrics)
```

### Train Model

```python
train_x_data = {'input_ids': np.array(trainX), 'attention_mask': np.array(train_attn_masks)}
test_x_data = {'input_ids': np.array(testX), 'attention_mask': np.array(test_attn_masks)}
```

```python
# Training the model with the prepared data
model.fit(
    train_x_data,
    trainY,
    validation_data=(test_x_data, testY),
    batch_size=16,
    epochs=3
)
```

```python
import tf_keras
from transformers import AutoTokenizer, TFAutoModelForSequenceClassification

# Load DeBERTa-v3 model and tokenizer
model_name = "microsoft/deberta-v3-base"
tokenizer = AutoTokenizer.from_pretrained(model_name, type_before_tokenization=True)

# Note the critical num_labels=2 fix
model = TFAutoModelForSequenceClassification.from_pretrained(
    model_name, 
    num_labels=2, 
    from_pt=True
)
```

```python
import re

def clean_text(text):
    if not isinstance(text, str):
        return ""
    
    # 1. Decode common broken UTF-8 patterns (like the 'üëç' emoji artifacts)
    try:
        # Re-encode to latin1 then decode properly to fix mojibake if present
        text = text.encode('latin1').decode('utf-8')
    except (UnicodeEncodeError, UnicodeDecodeError):
        pass
        
    # 2. Convert raw escaped strings (like '\r\n') into standard whitespace
    text = text.replace(r'\r', ' ').replace(r'\n', ' ').replace('\r', ' ').replace('\n', ' ')
    
    # 3. Strip out explicit remains of character encoding garbage artifacts
    text = re.sub(r'ü\S*', ' ', text) 
    
    # 4. Clean up any accidental double spacing or trailing gaps
    text = re.sub(r'\s+', ' ', text).strip()
    
    return text
```

```python
import tf_keras

# Calculate total training steps
batch_size = 16
epochs = 3  # Lowered epochs to prevent immediate overfitting
total_train_steps = (len(trainX) // batch_size) * epochs

# 1. Linear Learning Rate Decay Schedule
lr_schedule = tf_keras.optimizers.schedules.PolynomialDecay(
    initial_learning_rate=2e-5, # DeBERTa prefers lower initial learning rates than BERT
    decay_steps=total_train_steps,
    end_learning_rate=0.0,
    power=1.0
)

optimizer = tf_keras.optimizers.Adam(learning_rate=lr_schedule, epsilon=1e-08, clipnorm=1.0)
loss = tf_keras.losses.SparseCategoricalCrossentropy(from_logits=True)
metrics = [tf_keras.metrics.SparseCategoricalAccuracy('accuracy')]

model.compile(optimizer=optimizer, loss=loss, metrics=metrics)

# 2. Early Stopping Callback to save time and track the best weights
callbacks = [
    tf_keras.callbacks.EarlyStopping(
        monitor='val_loss', 
        patience=1, 
        restore_best_weights=True
    )
]

# Run training
model.fit(
    train_x_data,
    trainY,
    validation_data=(test_x_data, testY),
    batch_size=batch_size,
    epochs=epochs,
    callbacks=callbacks
)
```
